### 5.3 Normality Testing

The Shapiro-Wilk test is applied to examine whether numerical variables follow a normal distribution.

The hypotheses are:

* **H₀:** The data follows a normal distribution.
* **H₁:** The data does not follow a normal distribution.

The significance level is set at 0.05.

If the p-value is less than 0.05, the null hypothesis is rejected, indicating that the variable does not follow a normal distribution.

### 5.4 Correlation Analysis

Spearman's rank correlation is used to examine the relationship between numerical variables and the binary disease variable.

Spearman correlation measures the strength and direction of a monotonic relationship between two variables.

Its coefficient ranges from -1 to +1:

* +1 indicates a perfect positive relationship.
* -1 indicates a perfect negative relationship.
* 0 indicates no monotonic relationship.

This method is selected because the numerical variables do not satisfy the normality assumption.

### 5.5 Hypothesis Testing

Two statistical tests are used to examine relationships between patient characteristics and heart disease.

**A. Mann-Whitney U Test**

The Mann-Whitney U test is used to compare numerical variables between patients with and without heart disease.

Hypotheses:

* **H₀:** There is no statistically significant difference in the distributions of the selected variable between the two groups.
* **H₁:** There is a statistically significant difference between the two groups.

The test is applied to age, resting blood pressure, cholesterol, maximum heart rate, and ST depression.

**B. Chi-Square Test of Independence**

The Chi-square test is used to examine the association between categorical variables and heart disease.

Hypotheses:

* **H₀:** There is no significant association between the categorical variable and heart disease.
* **H₁:** There is a significant association between the categorical variable and heart disease.

The test is applied to sex, chest pain, fasting blood sugar, resting ECG, exercise-induced angina, slope, number of major vessels, and thalassemia-related results.

A significance level of 0.05 is used for both tests.

### 5.6 Logistic Regression

Logistic regression is used to predict the probability of heart disease.

Since the target variable contains two classes (0 and 1), logistic regression is suitable for this binary classification problem.

The logistic regression function is:

P(Y = 1) = 1 / (1 + e^(-z))

where:

z = β₀ + β₁X₁ + β₂X₂ + ... + βₙXₙ

Here:

* P(Y = 1) represents the predicted probability of heart disease.
* β₀ represents the intercept.
* β₁, β₂, ..., βₙ represent model coefficients.
* X₁, X₂, ..., Xₙ represent input variables.
* e represents the exponential constant.

The dataset is divided into training and testing sets using an 80:20 ratio. Stratified sampling is used to maintain class proportions.

StandardScaler is applied to scale the input features. The logistic regression model is trained using the training data and evaluated using the testing data.

### 5.7 Model Evaluation

* Accuracy

* Receiver Operating Characteristic – Area Under the Curve (ROC-AUC)

A confusion matrix is also generated to examine correct and incorrect predictions.

### 5.8 Risk Probability Estimation

The trained logistic regression model uses the `predict_proba()` function to estimate the probability of heart disease for a given patient.

For demonstration purposes, the predicted probability is divided into three project-defined categories:

* **Low Risk:** Probability below 30%.
* **Moderate Risk:** Probability from 30% to below 60%.
* **High Risk:** Probability of 60% or above.

These categories are created for academic demonstration and are not medically validated diagnostic thresholds.

```python
from scipy.stats import chi2_contingency

categorical_cols = [
    'sex',
    'cp',
    'fbs',
    'restecg',
    'exang',
    'slope',
    'ca',
    'thal'
]

print("Chi-square Test\n")

for col in categorical_cols:
    table = pd.crosstab(df[col], df['disease'])
    chi2, p, dof, expected = chi2_contingency(table)

    print(f"{col}:")
    print(f"  Chi-square = {chi2:.4f}")
    print(f"  df = {dof}")
    print(f"  p-value = {p:.6f}")
    print()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    roc_auc_score
)

# Features and target
X = df.drop(columns=['target', 'disease'])
y = df['disease']

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Standardize numerical variables
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Create logistic regression model
model = LogisticRegression(max_iter=1000)

# Train model
model.fit(X_train_scaled, y_train)

# Predictions
y_pred = model.predict(X_test_scaled)
y_prob = model.predict_proba(X_test_scaled)[:, 1]

# Evaluation
accuracy = accuracy_score(y_test, y_pred)
auc = roc_auc_score(y_test, y_prob)

print("Logistic Regression Results")
print("---------------------------")
print(f"Accuracy: {accuracy:.4f}")
print(f"ROC-AUC: {auc:.4f}")

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
import pandas as pd

def predict_heart_disease(
    age, sex, cp, trestbps, chol, fbs,
    restecg, thalach, exang, oldpeak,
    slope, ca, thal
):

    patient = pd.DataFrame([{
        'age': age,
        'sex': sex,
        'cp': cp,
        'trestbps': trestbps,
        'chol': chol,
        'fbs': fbs,
        'restecg': restecg,
        'thalach': thalach,
        'exang': exang,
        'oldpeak': oldpeak,
        'slope': slope,
        'ca': ca,
        'thal': thal
    }])

    # Apply the same scaling used during training
    patient_scaled = scaler.transform(patient)

    # Probability of disease
    probability = model.predict_proba(patient_scaled)[0][1]

    # Convert to percentage
    risk_percentage = probability * 100

    # Risk category
    if risk_percentage < 30:
        category = "Low"
    elif risk_percentage < 60:
        category = "Moderate"
    else:
        category = "High"

    print(f"Estimated Heart Disease Risk: {risk_percentage:.2f}%")
    print(f"Risk Category: {category}")

    return risk_percentage, category
```

```python
import gradio as gr
import pandas as pd

# --------------------------------------------------
# Risk prediction function
# --------------------------------------------------

def risk_predictor(
    age, sex, cp, trestbps, chol, fbs,
    restecg, thalach, exang, oldpeak,
    slope, ca, thal
):

    patient = pd.DataFrame([{
        'age': age,
        'sex': sex,
        'cp': cp,
        'trestbps': trestbps,
        'chol': chol,
        'fbs': fbs,
        'restecg': restecg,
        'thalach': thalach,
        'exang': exang,
        'oldpeak': oldpeak,
        'slope': slope,
        'ca': ca,
        'thal': thal
    }])

    # Apply the same scaler used during model training
    patient_scaled = scaler.transform(patient)

    # Predict probability
    probability = model.predict_proba(patient_scaled)[0][1]
    risk = probability * 100

    # Risk category
    if risk < 30:
        category = "LOW RISK"
    elif risk < 60:
        category = "MODERATE RISK"
    else:
        category = "HIGH RISK"

    return f"{risk:.2f}%", category


# --------------------------------------------------
# Gradio Interface
# --------------------------------------------------

with gr.Blocks(
    title="Heart Disease Risk Predictor"
) as app:

    gr.Markdown("""
    # ❤️ Heart Disease Risk Predictor

    ### Probability & Statistics Project

    This application uses a **Logistic Regression model**
    trained on the UCI Cleveland Heart Disease dataset
    to estimate the probability of heart disease.

    **Model Accuracy:** 86.89%
    **ROC-AUC:** 95.13%
    """)

    gr.Markdown("## 👤 Patient Information")

    with gr.Row():

        # LEFT COLUMN
        with gr.Column():

            age = gr.Number(
                label="Age",
                value=55,
                minimum=1,
                maximum=100
            )

            sex = gr.Dropdown(
                choices=[
                    ("Female", 0),
                    ("Male", 1)
                ],
                value=1,
                type="value",
                label="Sex"
            )

            cp = gr.Dropdown(
               choices=[
               ("Typical Angina", 1),
               ("Atypical Angina", 2),
               ("Non-anginal Pain", 3),
               ("Asymptomatic", 4)
               ],
                value=1,
               type="value",
               label="Chest Pain Type"
            )


            trestbps = gr.Number(
                label="Resting Blood Pressure (mm Hg)",
                value=140,
                minimum=50,
                maximum=250
            )

            chol = gr.Number(
                label="Serum Cholesterol (mg/dl)",
                value=250,
                minimum=50,
                maximum=700
            )

            fbs = gr.Dropdown(
                choices=[
                    ("No (≤ 120 mg/dl)", 0),
                    ("Yes (> 120 mg/dl)", 1)
                ],
                value=0,
                type="value",
                label="Fasting Blood Sugar"
            )

            restecg = gr.Dropdown(
                choices=[
                    ("Normal", 0),
                    ("ST-T Wave Abnormality", 1),
                    ("Left Ventricular Hypertrophy", 2)
                ],
                value=1,
                type="value",
                label="Resting ECG"
            )

        # RIGHT COLUMN
        with gr.Column():

            thalach = gr.Number(
                label="Maximum Heart Rate Achieved",
                value=150,
                minimum=50,
                maximum=250
            )

            exang = gr.Dropdown(
                choices=[
                    ("No", 0),
                    ("Yes", 1)
                ],
                value=0,
                type="value",
                label="Exercise-Induced Angina"
            )

            oldpeak = gr.Number(
                label="ST Depression (Oldpeak)",
                value=1.0,
                minimum=0,
                maximum=10
            )

            slope = gr.Dropdown(
              choices=[
                ("Upsloping", 1),
                ("Flat", 2),
                ("Downsloping", 3)
              ],
              value=2,
              type="value",
              label="ST Segment Slope"
            )

            ca = gr.Dropdown(
                choices=[
                    ("0 vessels", 0),
                    ("1 vessel", 1),
                    ("2 vessels", 2),
                    ("3 vessels", 3)
                ],
                value=0,
                type="value",
                label="Number of Major Vessels"
            )

            thal = gr.Dropdown(
                choices=[
                    ("Normal", 3),
                    ("Fixed Defect", 6),
                    ("Reversible Defect", 7)
                ],
                value=3,
                type="value",
                label="Thalassemia Result"
            )

    gr.Markdown("---")

    predict_button = gr.Button(
        "🔍 Predict Heart Disease Risk",
        variant="primary"
    )

    gr.Markdown("## 📊 Prediction Result")

    with gr.Row():

        risk_output = gr.Textbox(
            label="Estimated Heart Disease Risk",
            interactive=False
        )

        category_output = gr.Textbox(
            label="Risk Category",
            interactive=False
        )

    predict_button.click(
        fn=risk_predictor,
        inputs=[
            age, sex, cp, trestbps, chol, fbs,
            restecg, thalach, exang, oldpeak,
            slope, ca, thal
        ],
        outputs=[
            risk_output,
            category_output
        ]
    )

    gr.Markdown("""
    ---

    ### ⚠️ Disclaimer

    This application provides an **analytical risk estimate**
    based on a statistical machine-learning model and the
    UCI Cleveland Heart Disease dataset.

    **It is not a medical diagnosis and should not replace
    professional medical advice.**
    """)


# Launch application
app.launch(share=True)
```

## 8. Discussion

The statistical analysis provides insights into the relationship between patient characteristics and heart disease.

The Shapiro-Wilk test indicated that all five selected numerical variables deviate significantly from a normal distribution. Therefore, non-parametric methods were used for the final correlation and group comparison analyses.

The Spearman correlation results showed that ST depression has the strongest positive correlation with heart disease, while maximum heart rate has the strongest negative correlation among the numerical variables considered.

The Mann-Whitney U test identified statistically significant differences in all five numerical variables between patients with and without heart disease.

The Chi-square analysis showed significant associations for seven categorical variables. Fasting blood sugar was the only variable that did not show a statistically significant association.

The logistic regression model achieved 86.89% accuracy and a ROC-AUC score of 95.13%. Its disease-class recall was approximately 92.86%, showing that it correctly identified most disease-positive cases in the test dataset.

Overall, the findings demonstrate how descriptive statistics, correlation analysis, hypothesis testing, and predictive modelling can be combined to study heart disease data.


## 9. Conclusion

This project demonstrates the application of Probability and Statistics in analysing and predicting heart disease risk using the UCI Heart Disease Cleveland dataset.

Different statistical techniques, including descriptive statistics, the Shapiro-Wilk test, Spearman correlation, Mann-Whitney U test, and Chi-square test, were applied to examine the dataset and identify significant relationships.

The results revealed that several numerical and categorical variables are statistically associated with heart disease. The logistic regression model achieved an accuracy of 86.89% and a ROC-AUC score of 95.13% on the test dataset.

An interactive risk prediction interface was also developed to demonstrate how the trained model can estimate the probability of heart disease based on patient information.

The project highlights the importance of statistical analysis and probability-based modelling in understanding healthcare data. It also provides practical experience in data preprocessing, hypothesis testing, machine learning, and model evaluation.


## 10. Limitations

Although the project provides useful statistical insights, it has certain limitations:

1. **Limited Dataset Size:** The study uses only 303 records from the Cleveland subset of the UCI Heart Disease dataset.
2. **Limited Data Diversity:** The analysis is based on one dataset and may not represent all patient populations.
3. **Single Train-Test Split:** Model evaluation uses an 80:20 train-test split rather than extensive external validation.
4. **Feature Limitations:** The prediction model uses only the attributes available in the selected dataset.
5. **Risk Categories:** The low, moderate, and high probability categories are project-defined and have not been medically validated.
6. **Not a Diagnostic System:** The model is developed for academic purposes and must not be used as a replacement for professional medical advice or diagnosis.

Future work may include testing additional machine learning algorithms, using larger datasets, applying cross-validation, and evaluating the model on independent patient data.
